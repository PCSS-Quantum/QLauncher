# Using a custom Hamiltonian from `Qiskit`
In this tutorial we will go through a short pipeline of running QAOA (Quantum Approximate Optimization Algorithm) on a custom Hamiltonian with `QLauncher`. We will cover the following steps:
1) Defining a custom Hamiltonian
2) Wrapping it into a `QLauncher`‑compatible model
3) Initializing `QLauncher`'s QAOA instance
4) Executing QAOA with `QLauncher`
5) Summarizing the complete workflow

### 1. Defining a custom Hamiltonian
As an example we will build a Hamiltonian using `Qiskit`. We will use the representations of a sum of Pauli strings (I, Z, Y, X) with coefficients, stored in a memory‑efficient (sparse) format (`SparsePauliOp`). It is the standard way `Qiskit` represents Hamiltonians for simulation and optimization.

```python
from qiskit.quantum_info import SparsePauliOp

# Build Hamiltonian as Qiskit SparsePauliOp class
H = SparsePauliOp.from_list([
    ("II", 1.5),
    ("ZI", -1.0),
    ("ZZ", -0.5)
])
```

### 2. Wrapping it into a `QLauncher`‑compatible model
We can directly wrap our Hamiltonian into `QLauncher`’s built‑in Hamiltonian model. This model acts as a standardized model that `QLauncher` understands and can pass to any algorithm or backend that expects a Hamiltonian‑based problem. Look how simple is this in code!

```python
from qlauncher.base.models import Hamiltonian

H_ql = Hamiltonian(H)
```

Wrapping the `Qiskit` Hamiltonian into `QLauncher`'s Hamiltonian makes it possible to further translate the problem definition, for example into a QUBO or `Cirq` representation, using additional `QLauncher` functions.

### 3. Initializing `QLauncher`'s QAOA instance
We define the `algorithm` which is a part of `QLauncher`'s `Qiskit routines` and create `backend`. We can specify additional parameters for the QAOA algorithm. The parameter `p=2` sets the depth of the QAOA circuit, controlling how many alternating operator layers are applied. A deeper circuit can capture more complex correlations but requires more computational resources. The `optimization_method` argument specifies the classical optimizer used to tune QAOA’s variational parameters, while `max_evaluations` limits the number of objective function evaluations during the optimization process.

```python
from qlauncher.routines.qiskit import QAOA, QiskitBackend

algorithm = QAOA(p=2, optimization_method='COBYLA', max_evaluations=200)

backend = QiskitBackend('local_simulator')
```

### 4. Executing QAOA with `QLauncher`

With the algorithm and backend configured, the final step is to submit everything to `QLauncher` for execution. It acts as a unified interface that bundles together the problem Hamiltonian, optimization routine or any other algoritm and computing backend. This allows you to run the same model across different computational environments, algorithms or input instances with minimal code changes.

```python
from qlauncher import QLauncher
from pprint import pprint

result = QLauncher(H_ql, algorithm, backend).run()
pprint(vars(result))
```

With the results, we can for example extract the distribution and check which bitstrings have the best results

```python
import matplotlib.pyplot as plt

plt.bar(list(result.distribution.keys()), list(result.distribution.values()))
plt.show()
```

Result of running the `Launcher` with problem definition, algorithm and backend stores a lot of information apart from the best bitstring, energy and distribution which can be useful to track and assess the algorithm.

### 4. Summary
In few steps we adapted a custom Hamiltonian defined in `Qiskit` and run QAOA algorithm with `QLauncher`. But this is just a brief concrete example. Imagine how many algorithms and backends can be tested in a similar way and with the same initial Hamiltonian. `QLauncher` can simulate various architectures including simulated quantum annealing, photonic simulators as well as run the algorithms on real hardware including photonic and ion-trapped machines.
